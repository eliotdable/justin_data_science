```python
import numpy as np
import matplotlib.pyplot as plt

terms = ['extension', 'demolition', 'erection', 'change of use', 'variation',
         'alteration', 'application', 'installation', 'removal', 'amendment',
         'construction', 'redevelopment', 'replacement']

desc = df['description'].str.lower()
flags = pd.DataFrame({t: desc.str.contains(t, regex=False, na=False) for t in terms}).astype(int)
co = flags.T @ flags

# Drop the first row and last column; they would only hold the diagonal or blanks
co_plot = co.iloc[1:, :-1]
rows, cols = co_plot.index, co_plot.columns

# Hide everything above the old diagonal
mask = np.triu(np.ones(co_plot.shape, dtype=bool), k=1)
data = np.ma.masked_array(co_plot.values, mask)

fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(data, cmap='Blues')

ax.set_xticks(range(len(cols)), cols, rotation=45, ha='right')
ax.set_yticks(range(len(rows)), rows)
ax.spines[:].set_visible(False)

vmax = data.max()
for i in range(len(rows)):
    for j in range(i + 1):
        v = co_plot.iat[i, j]
        ax.text(j, i, f'{v:,}', ha='center', va='center', fontsize=8,
                color='white' if v > vmax * 0.6 else '#333333')

fig.colorbar(im, ax=ax, shrink=0.8, label='Descriptions containing both terms')
ax.set_title('Co-occurrence of terms in descriptions')
plt.tight_layout()
plt.show()
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt

categories = {
    'New build':           ['erection', 'construction', 'redevelopment'],
    'Extension': ['extension'],
    'Demolition/removal':  ['demolition', 'removal'],
    'Replacement/install': ['replacement', 'installation'],
    'Change':  ['change of use',  'alteration'],
    'Amendment/variation': ['variation', 'amendment', 'application'],
}

desc = df['description'].str.lower()

# A row belongs to a category if it contains any of that category's terms
cat_flags = pd.DataFrame({
    cat: desc.str.contains('|'.join(map(re.escape, words)), na=False)
    for cat, words in categories.items()
}).astype(int)

co = cat_flags.T @ cat_flags


def plot_cooccurrence(co, title):
    # Drop the diagonal, plus the first row and last column it leaves empty
    co_plot = co.iloc[1:, :-1]
    rows, cols = co_plot.index, co_plot.columns

    mask = np.triu(np.ones(co_plot.shape, dtype=bool), k=1)
    data = np.ma.masked_array(co_plot.values, mask)

    fig, ax = plt.subplots(figsize=(8, 6.5))
    im = ax.imshow(data, cmap='Blues')

    ax.set_xticks(range(len(cols)), cols, rotation=45, ha='right')
    ax.set_yticks(range(len(rows)), rows)
    ax.spines[:].set_visible(False)

    vmax = data.max()
    for i in range(len(rows)):
        for j in range(i + 1):
            v = co_plot.iat[i, j]
            ax.text(j, i, f'{v:,}', ha='center', va='center', fontsize=9,
                    color='white' if v > vmax * 0.6 else '#333333')

    fig.colorbar(im, ax=ax, shrink=0.8, label='Descriptions containing both')
    ax.set_title(title)
    plt.tight_layout()
    plt.show()


plot_cooccurrence(co, 'Co-occurrence of categories in descriptions')
```
